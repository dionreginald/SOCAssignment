```python
from flask import Flask, render_template, request, redirect, session, url_for
from functools import wraps
import sqlite3
from threading import Thread
import shutil
import os

app = Flask(__name__)
app.secret_key = 'cozycomfort123'
DB = "cozycomfort.db"

# If the uploaded DB is in /mnt/data, copy it to working directory once
if not os.path.exists(DB):
    shutil.copy("/mnt/data/cozycomfort.db", DB)
```

```python
@app.route("/", methods=["GET"])
def home():
    return render_template("index.html")  # Tailwind design page

@app.route("/select_user", methods=["POST"])
def select_user():
    session["selected_role"] = request.form["user_type"].lower()
    return redirect(url_for("login"))
```

```python
@app.route("/login", methods=["GET", "POST"])
def login():
    if request.method == "POST":
        username = request.form["username"].strip().lower()
        password = request.form["password"].strip()

        conn = sqlite3.connect(DB)
        c = conn.cursor()
        c.execute("SELECT role, password FROM users WHERE username = ?", (username,))
        result = c.fetchone()
        conn.close()

        if result:
            role, db_password = result
            selected_role = session.get("selected_role")

            if password != db_password:
                return render_template("login.html", error="Invalid password.", 
                                       selected_role=selected_role)

            if role == selected_role:
                session["username"] = username
                session["role"] = role
                return redirect(f"/{role}")
            else:
                return render_template("login.html", error=f"User '{username}' does not belong to {selected_role.capitalize()} role.",
                                       selected_role=selected_role)
        else:
            return render_template("login.html", error="Invalid username.", 
                                   selected_role=session.get("selected_role"))

    return render_template("login.html", selected_role=session.get("selected_role"))
```

```python
@app.route("/manufacturer")
def manufacturer_home():
    if session.get("role") != "manufacturer":
        return redirect("/")

    conn = sqlite3.connect(DB)
    c = conn.cursor()

    # Manufacturer stock display
    c.execute("""
        SELECT p.model, p.material, m.quantity
        FROM manufacturer_stock m
        JOIN products p ON m.product_id = p.id
    """)
    stock = c.fetchall()

    # Distributor stock requests
    c.execute("""
        SELECT r.id, p.model, p.material, r.quantity
        FROM stock_requests r
        JOIN products p ON r.product_id = p.id
        WHERE r.to_role = 'manufacturer' AND r.status = 'Pending'
    """)
    requests = c.fetchall()

    conn.close()
    return render_template("manufacturer_home.html", stock=stock, requests=requests)


@app.route("/manufacturer/add_product", methods=["GET", "POST"])
def add_product():
    if session.get("role") != "manufacturer":
        return redirect("/")

    conn = sqlite3.connect(DB)
    c = conn.cursor()

    if request.method == "POST":
        model = request.form.get("model", "").strip()
        material = request.form.get("material", "").strip()

        if model and material:
            try:
                c.execute("INSERT INTO products (model, material) VALUES (?, ?)", (model, material))
                conn.commit()
            except Exception as e:
                conn.close()
                return f"<h3 style='color:red'>Error adding product: {e}</h3>"

    c.execute("SELECT model, material FROM products")
    products = c.fetchall()
    conn.close()
    return render_template("manufacturer_add_product.html", products=products)


@app.route("/manufacturer/add_stock", methods=["GET", "POST"])
def add_stock():
    if session.get("role") != "manufacturer":
        return redirect("/")

    conn = sqlite3.connect(DB)
    c = conn.cursor()

    if request.method == "POST":
        product_id = request.form.get("product_id")
        quantity = request.form.get("quantity")

        if product_id and quantity:
            try:
                product_id = int(product_id)
                quantity = int(quantity)

                # Check if stock already exists
                c.execute("SELECT quantity FROM manufacturer_stock WHERE product_id=?", (product_id,))
                row = c.fetchone()

                if row:
                    c.execute("UPDATE manufacturer_stock SET quantity = quantity + ? WHERE product_id = ?", (quantity, product_id))
                else:
                    c.execute("INSERT INTO manufacturer_stock (product_id, quantity) VALUES (?, ?)", (product_id, quantity))

                conn.commit()
            except Exception as e:
                conn.close()
                return f"<h3 style='color:red'>Error adding stock: {e}</h3>"

    # Products for dropdown
    c.execute("SELECT id, model, material FROM products")
    products = c.fetchall()

    # Current stock
    c.execute("""
        SELECT p.model, p.material, m.quantity
        FROM manufacturer_stock m
        JOIN products p ON m.product_id = p.id
    """)
    stock = c.fetchall()

    conn.close()
    return render_template("manufacturer_add_stock.html", products=products, stock=stock)
```

```python
@app.route("/distributor", methods=["GET"])
@login_required("distributor")
def distributor_home():
    conn = sqlite3.connect(DB)
    c = conn.cursor()

    # Distributor's stock
    c.execute("""
        SELECT p.model, p.material, d.quantity
        FROM distributor_stock d
        JOIN products p ON d.product_id = p.id
    """)
    stock = c.fetchall()

    # Manufacturer stock (view-only)
    c.execute("""
        SELECT p.model, p.material, m.quantity
        FROM manufacturer_stock m
        JOIN products p ON m.product_id = p.id
    """)
    mstock = c.fetchall()

    # Seller requests pending for this distributor
    c.execute("""
        SELECT r.id, p.model, p.material, r.quantity
        FROM stock_requests r
        JOIN products p ON r.product_id = p.id
        WHERE r.to_role = 'distributor' AND r.status = 'Pending'
    """)
    seller_requests = c.fetchall()

    conn.close()
    return render_template(
        "distributor_home.html",
        stock=stock,
        mstock=mstock,
        seller_requests=seller_requests
    )

# ----------------------------------------
# Distributor Request - Send stock request to Manufacturer
# ----------------------------------------
@app.route("/distributor/request", methods=["GET", "POST"])
@login_required("distributor")
def distributor_request():
    conn = sqlite3.connect(DB)
    c = conn.cursor()

    if request.method == "POST":
        try:
            product_id = int(request.form["product_id"])
            quantity = int(request.form["quantity"])

            # Insert request
            c.execute("""
                INSERT INTO stock_requests (from_role, to_role, product_id, quantity, status)
                VALUES ('distributor', 'manufacturer', ?, ?, 'Pending')
            """, (product_id, quantity))
            conn.commit()
            conn.close()
            return redirect(url_for("distributor_home"))

        except Exception as e:
            conn.close()
            return f"<h3 style='color:red'>Error while requesting stock: {e}</h3>"

    # GET request – show product dropdown
    c.execute("SELECT id, model, material FROM products")
    products = c.fetchall()
    conn.close()
    return render_template("distributor_request.html", products=products)
```

```python
@app.route("/seller", methods=["GET", "POST"])
def seller():
    if session.get("role") != "seller":
        return redirect("/")
    conn = sqlite3.connect(DB)
    c = conn.cursor()

    if request.method == "POST":
        product_id = int(request.form["product_id"])
        qty = int(request.form["quantity"])
        c.execute("INSERT INTO stock_requests (from_role, to_role, product_id, quantity, status) VALUES ('seller', 'distributor', ?, ?, 'Pending')", (product_id, qty))
        conn.commit()

    c.execute("SELECT p.id, p.model, p.material FROM products p")
    products = c.fetchall()
    c.execute("SELECT p.model, p.material, s.quantity FROM seller_stock s JOIN products p ON s.product_id=p.id")
    stock = c.fetchall()
    conn.close()
    return render_template("seller.html", products=products, stock=stock)
```

```python
@app.route("/approve/<int:req_id>")
def approve(req_id):
    conn = sqlite3.connect(DB)
    c = conn.cursor()

    # Fetch the stock request
    c.execute("SELECT * FROM stock_requests WHERE id=?", (req_id,))
    req = c.fetchone()

    if not req:
        conn.close()
        return "<h3 style='color:red'>Invalid request ID.</h3>"

    _, from_role, to_role, product_id, qty, status = req

    try:
        if to_role == "manufacturer":
            # Approving distributor's request → Manufacturer reduces, distributor increases
            c.execute("SELECT quantity FROM manufacturer_stock WHERE product_id=?", (product_id,))
            row = c.fetchone()

            if row and row[0] >= qty:
                c.execute("UPDATE manufacturer_stock SET quantity = quantity - ? WHERE product_id=?", (qty, product_id))

                # Check if distributor already has the product
                c.execute("SELECT quantity FROM distributor_stock WHERE product_id=?", (product_id,))
                if c.fetchone():
                    c.execute("UPDATE distributor_stock SET quantity = quantity + ? WHERE product_id=?", (qty, product_id))
                else:
                    c.execute("INSERT INTO distributor_stock (product_id, quantity) VALUES (?, ?)", (product_id, qty))

                c.execute("UPDATE stock_requests SET status = 'Approved' WHERE id = ?", (req_id,))
                conn.commit()
            else:
                conn.close()
                return "<h3 style='color:red'>Insufficient stock in Manufacturer.</h3>"

        elif to_role == "distributor":
            # Approving seller's request → Distributor reduces, seller increases
            c.execute("SELECT quantity FROM distributor_stock WHERE product_id=?", (product_id,))
            row = c.fetchone()

            if row and row[0] >= qty:
                c.execute("UPDATE distributor_stock SET quantity = quantity - ? WHERE product_id=?", (qty, product_id))

                # Check if seller already has the product
                c.execute("SELECT quantity FROM seller_stock WHERE product_id=?", (product_id,))
                if c.fetchone():
                    c.execute("UPDATE seller_stock SET quantity = quantity + ? WHERE product_id=?", (qty, product_id))
                else:
                    c.execute("INSERT INTO seller_stock (product_id, quantity) VALUES (?, ?)", (product_id, qty))

                c.execute("UPDATE stock_requests SET status = 'Approved' WHERE id = ?", (req_id,))
                conn.commit()
            else:
                conn.close()
                return "<h3 style='color:red'>Insufficient stock in Distributor.</h3>"

    except Exception as e:
        conn.rollback()
        conn.close()
        return f"<h3 style='color:red'>Error during approval: {e}</h3>"

    conn.close()
    return redirect(f"/{to_role}")
```

```python
@app.route("/manufacturer/deny/<int:request_id>")
def deny_request_manufacturer(request_id):
    # Ensure only manufacturer can access
    if session.get("role") != "manufacturer":
        return redirect("/")

    try:
        conn = sqlite3.connect(DB)
        c = conn.cursor()

        # Check if the request exists and is intended for the manufacturer
        c.execute("SELECT to_role FROM stock_requests WHERE id = ?", (request_id,))
        row = c.fetchone()

        if not row:
            conn.close()
            return "<h3 style='color:red'>Request not found.</h3>"

        if row[0] != "manufacturer":
            conn.close()
            return "<h3 style='color:red'>Unauthorized denial attempt.</h3>"

        # Update status to Denied
        c.execute("UPDATE stock_requests SET status = 'Denied' WHERE id = ?", (request_id,))
        conn.commit()
        conn.close()
        return redirect("/manufacturer")

    except Exception as e:
        return f"<h3 style='color:red'>Error while denying request: {e}</h3>"
```
